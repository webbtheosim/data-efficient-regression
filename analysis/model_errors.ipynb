```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('./data/errors.csv', header=None)
df.columns = ['task', 'model', 'seed', 'corr']
df = df[df['task'].isin(['esol', 'lipo', 'qm9_cv', 'polymer_tg'])]
```

```python
summary = (
    df.groupby(["task", "model"], as_index=False)
      .agg(
          corr_mean=("corr", "mean"),
          corr_sem=("corr", "sem"),
      )
)
reference = (
    summary[summary["model"] == "gp_ard"]
    .rename(columns={
        "corr_mean": "ref_mean",
        "corr_sem": "ref_sem",
    })
    [["task", "ref_mean", "ref_sem"]]
)
summary = summary.merge(reference, on="task")
summary["ratio"] = summary["corr_mean"] / summary["ref_mean"]
summary["ratio_sem"] = (
    summary["ratio"]
    * np.sqrt(
        (summary["corr_sem"] / summary["corr_mean"])**2
        + (summary["ref_sem"] / summary["ref_mean"])**2
    )
)
print(summary)
```
